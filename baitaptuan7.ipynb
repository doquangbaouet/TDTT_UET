## W7A1 Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):


```python
def binary_search(lst, target):
    low = 0 
    high = len(lst) - 1 
    
    while low <= high: 
        mid = (low + high)//2 
        if lst[mid] == target:
            return mid 
        elif lst[mid] < target:
            low = mid + 1 
        else:
            high = mid - 1 
    return -1 
lst = list(map(int,input().split()))
n = int(input())
print(binary_search(lst, n))

```

## W7A2 
Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) -
> int:


```python
def count_occurrences(arr, x):
    cnt = 0 
    for n in arr:
        if n == x: 
            cnt += 1 
    return cnt 
arr = list(map(int, input().split()))
x = int(input())
print(count_occurrences(arr,x))
    
```

## W7A3 Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort);Sắp xếp chọn (Selection
Sort)



```python
## bubble sort 
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1],arr[j]
    return arr
arr = list(map(int, input().split()))
print(bubble_sort(arr))
```

```python
## insertion sort
def insertion_sort(arr):
    for i in range(len(arr)):
        min_index = i 
        for j in range(i+1,len(arr)):
            if arr[j] < arr[min_index]:
                min_index = j 
        arr[i],arr[min_index] = arr[min_index], arr[i]
    return arr 
arr = list(map(int, input().split()))
print(insertion_sort(arr))
```

## W7A4 Viết hàm tìm phần tử có số
lần xuất hiện nhiều nhất
trong danh sách. Nếu có
nhiều phần tử cùng số lần,
trả về phần tử xuất hiện
trước.


```python
def find_most_frequent(arr):
    counts = {}
    for x in arr:
        counts[x] = counts.get(x, 0) + 1
    max_count = 0
    result = arr[0]

    for x in arr:
        if counts[x] > max_count:
            max_count = counts[x]
            result = x
    return result, max_count

arr = list(map(int, input().split()))
val, count = find_most_frequent(arr)
print(f"{val} xuất hiện nhiều nhất, {count} lần")

```

## W7A5 
Cho danh sách các số
nguyên và một số X, đếm
tất cả các cặp (a, b) sao
cho a + b = X. c

```python
def count_pairs(arr, x):
    seen = {}
    count = 0
    for num in arr:
        target = x - num
        if target in seen:
            count += seen[target]
        seen[num] = seen.get(num, 0) + 1
            
    return count


arr = list(map(int, input("Nhập danh sách: ").split()))
x = int(input("Nhập số X: "))
print(count_pairs(arr, x))
```

## W7A6 Bạn được cung cấp một
danh sách các danh sách
lồng nhau, vd [[1, 2, 3], [4,
5], [6, 7, 8, 9]], hãy làm
phẳng danh sách


```python
def flatten_list(nested_list):
    result = []
    for sublist in nested_list:
        for item in sublist:
            result.append(item)
    return result

raw_input = input()
nested_list = eval(raw_input)

print(flatten_list(nested_list))
```

## W7A7 Cho một danh sách các số
nguyên nums, hãy tìm độ
dài của dãy con tăng
nghiêm ngặt dài nhất
(không nhất thiết phải liền kề).


```python
def length_of_lis(nums):
    if not nums:
        return 0
    
    n = len(nums)
    dp = [1] * n
    
    for i in range(1, n):
        for j in range(i):
            if nums[i] > nums[j]:
                if dp[j] + 1 > dp[i]:
                    dp[i] = dp[j] + 1
                    
    return max(dp)

nums = eval(input())
print(length_of_lis(nums))
```

## W7A8 Bạn được cung cấp: một
danh sách các khoảng thời
gian interval = [[l1, r1], [l2,
r2], ...] và một danh sách
các truy vấn queries = [q1,
q2, ...]
Với mỗi truy vấn q, hãy tìm
độ dài nhỏ nhất của
khoảng thời gian [li, ri] sao
cho li ≤ q ≤ ri.
Nếu không tồn tại khoảng
thời gian nào như vậy, hãy
trả về -1 cho truy vấn đó.
solve in < O(m.n), hint:
binary search in the
queries list


```python
def solve_intervals(intervals, queries):
    
    data = []
    for l, r in intervals:
        data.append((r - l + 1, l, r))
    
  
    data.sort()
    
    results = []
    for q in queries:
        found = False
        for length, l, r in data:
            if l <= q <= r:
                results.append(length)
                found = True
                break
        if not found:
            results.append(-1)
    return results

intervals = eval(input())
queries = eval(input())

print(solve_intervals(intervals, queries))
```

## W7A9 Cho một danh sách các số
nums, với mỗi phần tử
nums[i], hãy đếm xem có
bao nhiêu phần tử bên phải
nhỏ hơn nums[i].
Trả về danh sách các số
đếm.
Efficient solution: using
heap or binary search tree


```python
def count_smaller_right(nums):
    
    class Node:
        def __init__(self, val):
            self.val = val
            self.left = None
            self.right = None
            self.count = 0  
            self.size = 1  
    def insert(root, val):
        count = 0
        if not root:
            return Node(val), 0
        
        if val < root.val:
            root.count += 1
            root.left, c = insert(root.left, val)
            count = c
        else:
            root.right, c = insert(root.right, val)
            count = root.count + (1 if val > root.val else 0) + c
        
        root.size = 1 + (root.left.size if root.left else 0) + (root.right.size if root.right else 0)
        return root, count

    res = []
    root = None
  
    for i in range(len(nums) - 1, -1, -1):
        root, count = insert(root, nums[i])
        res.append(count)
    
    return res[::-1]


nums = eval(input())
print(count_smaller_right(nums))
```
