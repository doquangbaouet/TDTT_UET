## w7A16 Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.


```python
arr = eval(input())

counts = {}
for x in arr:
    if x in counts:
        counts[x] = counts[x] + 1
    else:
        counts[x] = 1

y = 0
for x in counts:
    count = counts[x]
    if count < x:
        y = y + count
    elif count > x:
        y = y + (count - x)

print(y)
```

## W7A17 
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
nums = eval(input())
n = len(nums)

candidate = None
count = 0

for num in nums:
    if count == 0:
        candidate = num
        count = 1
    elif num == candidate:
        count = count + 1
    else:
        count = count - 1

check_count = 0
for num in nums:
    if num == candidate:
        check_count = check_count + 1

if check_count > n / 2:
    print(candidate)
```

## W7A18 Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; ; 

```python
nums = eval(input())

evens = []
odds = []

for x in nums:
    if x % 2 == 0:
        evens.append(x)
    else:
        odds.append(x)

evens.sort()
odds.sort(reverse=True)

result = evens + odds

print(result)
```

## W7A19 Merge 2 mảng đã sắp xếp (có thể tăng dần, có thể giảm dần) mà không
dùng bất cứ thuật toán sắp xếp nào


```python
nums1 = eval(input())
nums2 = eval(input())


if nums1[0] > nums1[-1]:
    nums1.reverse()
if nums2[0] > nums2[-1]:
    nums2.reverse()


result = []
i = 0
j = 0

while i < len(nums1) and j < len(nums2):
    if nums1[i] < nums2[j]:
        result.append(nums1[i])
        i = i + 1
    else:
        result.append(nums2[j])
        j = j + 1


while i < len(nums1):
    result.append(nums1[i])
    i = i + 1
while j < len(nums2):
    result.append(nums2[j])
    j = j + 1

print(result)
```

## W7A22 
Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
n = int(input())
names = []
for i in range(n):
    names.append(input())

counts = {}
for name in names:
    if name in counts:
        counts[name] = counts[name] + 1
    else:
        counts[name] = 1

best_name = ""
max_count = 0

for name in counts:
    if counts[name] > max_count:
        max_count = counts[name]
        best_name = name
    elif counts[name] == max_count:
        if best_name == "" or name < best_name:
            best_name = name

print(best_name)
```

## W7A24 Cho xâu kí tự, xóa các từ giống nhau liền kề, chỉ giữ lại từ đầu tiên

```python
s = input().split()

if not s:
    print("")
else:
    result = [s[0]]
    for i in range(1, len(s)):
        if s[i] != s[i - 1]:
            result.append(s[i])
    
    print(" ".join(result))
```

## W7A25 Tìm số có đúng k chữ số lớn nhất có thể trong một số có n chữ số (coi
đầu vào là một số rất lớn - str). Không được hoán đổi vị trí các chữ số


```python
n_str = input()
k = int(input())

result = ""
start = 0
end = len(n_str) - k

for i in range(k):
    max_digit = -1
    best_idx = -1
    
    for j in range(start, end + 1):
        if int(n_str[j]) > max_digit:
            max_digit = int(n_str[j])
            best_idx = j
            
    result = result + str(max_digit)
    start = best_idx + 1
    end = end + 1

print(result)
```

## W7A26 Tách các số có trong một đoạn văn bản, tính tổng và tính trung bình các
số đó

```python
text = input()

numbers = []
current_num = ""

for char in text:
    if char.isdigit():
        current_num = current_num + char
    else:
        if current_num != "":
            numbers.append(int(current_num))
            current_num = ""
if current_num != "":
    numbers.append(int(current_num))

if not numbers:
    print("Khong tim thay so")
else:
    total = 0
    for num in numbers:
        total = total + num
    
    average = total / len(numbers)
    
    print(total)
    print(average)
```

## W7A27 Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s = input()
stack = []
mapping = {')': '(', '}': '{', ']': '['}
is_valid = True

for char in s:
    if char in "({[":
        stack.append(char)
    else:
        if not stack or stack.pop() != mapping[char]:
            is_valid = False
            break

if is_valid and not stack:
    print("Hop le")
else:
    print("Khong hop le")
```

## W7A28 
Bà̀i toán Tìm dãy các hợp số dài nhất liên tiếp không chứa số nguyên
nào <= N (N là số nguyên dương cho trước). Nếu có nhiều dãy cùng độ
dài thì lấy dãy nhỏ nhất (gặp đầu tiên từ trái qua)


```python
def is_composite(num):
    if num < 4:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return True
    return False

N = int(input())
M = int(input()) 

max_len = 0
best_start = -1
current_len = 0
current_start = -1

for i in range(N + 1, M + 1):
    if is_composite(i):
        if current_len == 0:
            current_start = i
        current_len = current_len + 1
    else:
        if current_len > max_len:
            max_len = current_len
            best_start = current_start
        current_len = 0

if current_len > max_len:
    max_len = current_len
    best_start = current_start

if best_start != -1:
    print(best_start, max_len)
```

## W7A29 Tìm Dãy con tăng dần liên tục dài nhất của một danh sách nhập vào từ
bàn phím. Lấy dãy đầu tiên nếu có nhiều dãy có số phần tử bằng nhau

```python
nums = eval(input())

if not nums:
    print([])
else:
    max_len = 1
    best_start = 0
    current_len = 1
    current_start = 0

    for i in range(1, len(nums)):
        if nums[i] > nums[i - 1]:
            current_len = current_len + 1
        else:
            if current_len > max_len:
                max_len = current_len
                best_start = current_start
            current_len = 1
            current_start = i

    if current_len > max_len:
        max_len = current_len
        best_start = current_start

    print(nums[best_start : best_start + max_len])
```

## W7A30 Tìm Dãy con tăng dần KHÔNG liên tục dài nhất của một danh sách nhập
vào từ bàn phím 


```python
nums = eval(input())
n = len(nums)
if n == 0:
    print([])
else:
    dp = [1] * n
    prev = [-1] * n

    for i in range(1, n):
        for j in range(0, i):
            if nums[i] > nums[j]:
                if dp[j] + 1 > dp[i]:
                    dp[i] = dp[j] + 1
                    prev[i] = j

    max_len = 0
    last_idx = -1
    for i in range(n):
        if dp[i] > max_len:
            max_len = dp[i]
            last_idx = i

    result = []
    while last_idx != -1:
        result.append(nums[last_idx])
        last_idx = prev[last_idx]
    
    print(result[::-1])
```
